### 基于用户的协同过滤

1.找到与目标用户兴趣相似的用户集合

2.选出这些相似用户喜欢的且目标用户未关注的物品，把他们推荐给目标用户

Jaccard相似系数
$$
J(A,B)=\frac{|A\bigcap B|}{|A\bigcup B|}
$$

余弦相似度
$$
cos(A,B)=\frac{|A\bigcap B|}{\sqrt{|A|×|B|}}
$$

|用户|物品a|物品b|物品c|物品d|物品e|
|:----:|:----:|:----:|:----:|:----:|:----:|
|A|3|4|-|3.5|-|
|B|4|-|4.5|-|3.5|
|C|-|3.5|-|-|3|
|D|-|4|-|3.5|3|

以余弦相似度为例：  
$$C与A的相似度：w_{CA}=\frac{1}{\sqrt{2*3}}=\frac{1}{\sqrt{6}}$$
$$C与B的相似度：w_{CB}=\frac{1}{\sqrt{2*3}}=\frac{1}{\sqrt{6}}$$
$$C与D的相似度：w_{CD}=\frac{2}{\sqrt{2*3}}=\frac{1}{\sqrt{6}}$$ 



用户C对物品a的感兴趣程度为$P(C,a)$：
$$P(C,a)=w_{CA}×3.0+w_{CB}×4.0+w_{CD}×0=2.858$$

同理：
$$P(C,c)=w_{CA}×0+w_{CB}×4.5+w_{CD}×0=1.837$$
$$P(C,d)=w_{CA}×3.5+w_{CB}×0+w_{CD}×3.5=4.287$$

因此，推荐顺序：$d>a>c$

### 基于内容的协同过滤

物品相似度：
喜欢物品a的用户记作集合$w_1$  
喜欢物品b的用户记作集合$w_2$  
定义交集$v=w_1 \bigcap w_2$  
$$sim(a,b)=\frac{|v|}{\sqrt{|w_1|*|w_2|}}$$
